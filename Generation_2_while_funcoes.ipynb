```python
# como funciona o while

limite = int(input("informe o numero limite: "))
inicio = 0

while inicio <= limite:
  print(inicio)
  inicio = inicio + 1
```

```python
#controle com while
senha_correta = "abc123"
tentativas = 0

while tentativas < 3:
  senha = input("digite sua senha: ")
  if senha == senha_correta:
    print("Bem vindo ao sistema")
    break
  else:
    print("senha incorreta")
    tentativas = tentativas + 1
    print(f"numero de tentativas = {tentativas}")
    if tentativas == 3:
      print("tentativas esgotadas")
```

```python
# while true

while True:
  comando = input("para sair digite SAIR ")
  if comando.lower() == "sair":
    print("ate mais")
    break
  else:
    print("continuando o sistema")
```

```python
# Vamos criar um sistema que le um algoritmo.
# No Brasil a temperatura é Celcius (°C)
# No EUA é Fahrenheit (°F)
# Mas o certo é K - Kelvin

def celcius_f(temperatura):       # f = (valor em °C * 1.8) + 32
  f = (temperatura * 1.8) + 32
  return f                        # dentro de uma funcao usamos o comando return

def celcius_k(temperatura):
  k = temperatura + 273
  return k
```

```python
temperatura = float(input("Informe a temperatura: "))
print(celcius_f(temperatura))
print(celcius_k(temperatura))
```
